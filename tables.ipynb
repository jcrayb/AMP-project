```python
import yfinance as yf

all_stocks = ["NVDA", "AAPL", 'GOOG', "AVGO", "WMT", "JPM", "LLY", "XOM", "KO", "LMT", "F"]
```

```python
info = [yf.Ticker(s).info for s in all_stocks]
```

```python
info
```

```python
print(info[0].keys())
```

```python
print("""
\\begin{table}[!h]
    \\centering
    \\begin{tabular}{|c|c|}""")

print("\\hline Company & Sector \\\\ \\hline")

for i, t in enumerate(info):
    print(f"{t['longName']} & {t['sector']} ({t['industry']}) \\\\ \\hline" )

print("""\\hline   
    \\end{tabular}
\\end{table}
""")
```
